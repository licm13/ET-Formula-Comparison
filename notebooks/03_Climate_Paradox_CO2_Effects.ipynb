## 🔬 核心概念：CO2的双重效应 / Core Concept: Dual Effects of CO2

### 效应1：温室效应（促进蒸发）/ Effect 1: Greenhouse Effect (Promotes Evaporation)

```
CO2 ↑ → 全球变暖 → 温度 ↑ → 饱和水汽压 ↑ → ET ↑
CO2 ↑ → Global warming → Temperature ↑ → Saturation vapor pressure ↑ → ET ↑
```

### 效应2：气孔效应（抑制蒸发）/ Effect 2: Stomatal Effect (Suppresses Evaporation)

```
CO2 ↑ → 气孔部分关闭 → 气孔阻力 ↑ → 蒸腾 ↓ → ET ↓
CO2 ↑ → Stomata partially close → Stomatal resistance ↑ → Transpiration ↓ → ET ↓
```

### 植物为什么要关闭气孔？ / Why Do Plants Close Stomata?

植物进行光合作用时：

During photosynthesis, plants:

1. **吸收CO2** → 需要打开气孔 / **Absorb CO2** → Need to open stomata
2. **失去水分** → 蒸腾作用 / **Lose water** → Transpiration

这是一个**权衡 (trade-off)**：

This is a **trade-off**:

$$
\text{水分利用效率 / WUE} = \frac{\text{CO2吸收 / CO2 uptake}}{\text{水分损失 / Water loss}}
$$

**当CO2浓度高时 / When CO2 is abundant:**
- 植物可以"少呼吸"（部分关闭气孔）也能获得足够的CO2
- Plants can "breathe less" (partially close stomata) and still get enough CO2
- 结果：省水！💧
- Result: Save water! 💧

## 📐 数学建模：气孔导度 / Mathematical Modeling: Stomatal Conductance

### 经典关系 / Classic Relationship

$$
g_s \propto \frac{1}{\sqrt{C_a}}
$$

其中 / Where:
- $g_s$: 气孔导度 / Stomatal conductance (m/s)
- $C_a$: 大气CO2浓度 / Atmospheric CO2 concentration (ppm)

### PM-CO2 模型 / PM-CO2 Model

**原始PM方程中的表面阻力 / Surface resistance in original PM equation:**

$$
r_s = \text{constant} = 70 \, \text{s/m}
$$

**PM-CO2 模型中 / In PM-CO2 model:**

$$
r_s(CO2) = r_{s,\text{ref}} \times \sqrt{\frac{CO2}{CO2_{\text{ref}}}}
$$

其中 / Where:
- $CO2_{\text{ref}} = 380$ ppm（参考值）/ Reference CO2
- $r_{s,\text{ref}} = 70$ s/m（参考阻力）/ Reference resistance

**物理意义 / Physical meaning:**
- CO2从380 ppm → 760 ppm（翻倍）
- $r_s$ 增加 $\sqrt{2} \approx 1.41$ 倍
- 气孔导度 $g_s = 1/r_s$ 减少约30%！

### 实验1：CO2浓度的影响 / Experiment 1: Effect of CO2 Concentration

````python
# 固定气象条件 / Fixed meteorological conditions
T = 25.0
RH = 60.0
u2 = 2.5
Rn = 15.0

# CO2浓度范围（历史到未来）/ CO2 range (historical to future)
co2_levels = np.array([280, 380, 420, 550, 700, 900, 1200])  # ppm
co2_labels = ['Pre-industrial\n工业革命前\n(1750)', 
              'Reference\n参考\n(2000)', 
              'Current\n当前\n(2023)',
              'RCP4.5\n中等排放\n(2070)',
              'RCP6.0\n高排放\n(2080)',
              'RCP8.5\n极端排放\n(2090)',
              'Extreme\n极端\n(2150?)']

# 计算两种模型的ET / Calculate ET using both models
et_traditional = []
et_co2aware = []

for co2 in co2_levels:
    # 传统PM模型（不考虑CO2）/ Traditional PM (ignoring CO2)
    et_trad = penman_monteith(
        temperature=T,
        relative_humidity=RH,
        wind_speed=u2,
        net_radiation=Rn
    )
    et_traditional.append(et_trad)
    
    # CO2感知PM模型 / CO2-aware PM model
    et_co2 = pm_co2_aware(
        temperature=T,
        relative_humidity=RH,
        wind_speed=u2,
        net_radiation=Rn,
        co2=co2
    )
    et_co2aware.append(et_co2)

et_traditional = np.array(et_traditional)
et_co2aware = np.array(et_co2aware)

# 计算差异 / Calculate difference
difference = et_traditional - et_co2aware
percent_diff = (difference / et_traditional) * 100

# 绘制结果 / Plot results
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# 左图：绝对值比较 / Left: Absolute comparison
x = np.arange(len(co2_levels))
width = 0.35

bars1 = ax1.bar(x - width/2, et_traditional, width, label='Traditional PM\n传统PM', color='steelblue', alpha=0.8)
bars2 = ax1.bar(x + width/2, et_co2aware, width, label='PM-CO2\nCO2感知', color='forestgreen', alpha=0.8)

ax1.set_xlabel('CO2 Level / CO2水平', fontsize=12, fontweight='bold')
ax1.set_ylabel('PET (mm/day)', fontsize=12, fontweight='bold')
ax1.set_title('Traditional PM vs. CO2-aware PM\n传统PM vs. CO2感知PM', fontsize=14, fontweight='bold')
ax1.set_xticks(x)
ax1.set_xticklabels([f'{co2}\nppm' for co2 in co2_levels], fontsize=9)
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3, axis='y')

# 标注数值 / Annotate values
for i, (v1, v2) in enumerate(zip(et_traditional, et_co2aware)):
    ax1.text(i - width/2, v1 + 0.1, f'{v1:.2f}', ha='center', fontsize=9, fontweight='bold')
    ax1.text(i + width/2, v2 + 0.1, f'{v2:.2f}', ha='center', fontsize=9, fontweight='bold')

# 右图：差异百分比 / Right: Percentage difference
colors = ['green' if p < 5 else 'orange' if p < 15 else 'red' for p in percent_diff]
bars = ax2.bar(x, percent_diff, color=colors, alpha=0.7, edgecolor='black', linewidth=1.5)

ax2.set_xlabel('CO2 Level / CO2水平', fontsize=12, fontweight='bold')
ax2.set_ylabel('PET Overestimation / PET高估 (%)', fontsize=12, fontweight='bold')
ax2.set_title('Error by Ignoring CO2 Effects\n忽略CO2效应导致的误差', fontsize=14, fontweight='bold')
ax2.set_xticks(x)
ax2.set_xticklabels([f'{co2}\nppm' for co2 in co2_levels], fontsize=9)
ax2.axhline(y=0, color='black', linestyle='--', linewidth=1)
ax2.grid(True, alpha=0.3, axis='y')

# 标注百分比 / Annotate percentages
for i, p in enumerate(percent_diff):
    ax2.text(i, p + 0.5, f'{p:.1f}%', ha='center', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()

# 打印详细结果 / Print detailed results
print("\n📊 详细比较 / Detailed Comparison")
print("="*80)
print(f"{'CO2 (ppm)':<12} {'Traditional PM':<15} {'PM-CO2':<15} {'Difference':<15} {'Error (%)':<10}")
print("-"*80)
for i, co2 in enumerate(co2_levels):
    print(f"{co2:<12} {et_traditional[i]:<15.3f} {et_co2aware[i]:<15.3f} "
          f"{difference[i]:<15.3f} {percent_diff[i]:<10.1f}")

print("\n💡 关键发现 / Key Findings:")
print(f"  • 在当前CO2水平(420 ppm)，传统模型高估ET约 {percent_diff[2]:.1f}%")
print(f"  • At current CO2 (420 ppm), traditional model overestimates ET by ~{percent_diff[2]:.1f}%")
print(f"\n  • 在RCP8.5情景(900 ppm)，高估达到 {percent_diff[5]:.1f}%！")
print(f"  • Under RCP8.5 scenario (900 ppm), overestimation reaches {percent_diff[5]:.1f}%!")
print(f"\n🌍 这对全球水循环预测有重大影响！")
print(f"🌍 This has major implications for global water cycle predictions!")
````

## 🌍 实验2：2020-2100年情景模拟 / Experiment 2: 2020-2100 Scenario Simulation

### 气候变化情景 / Climate Change Scenarios

**IPCC RCP情景 / IPCC RCP Scenarios:**

- **RCP2.6**: 严格减排（巴黎协定目标）/ Aggressive mitigation (Paris Agreement)
- **RCP4.5**: 中等减排 / Moderate mitigation
- **RCP8.5**: 无减排（最坏情况）/ No mitigation (worst case)

````python
# 时间序列 / Time series
years = np.arange(2020, 2101, 5)  # 2020-2100，每5年 / Every 5 years

# CO2情景（根据IPCC RCP）/ CO2 scenarios (based on IPCC RCP)
co2_rcp26 = 420 + (420 - 420) * (years - 2020) / 80 + 10 * np.log(1 + (years - 2020) / 80)
co2_rcp45 = 420 + (550 - 420) * (years - 2020) / 80
co2_rcp85 = 420 + (900 - 420) * (years - 2020) / 80

# 温度异常（相对于2020年）/ Temperature anomaly (relative to 2020)
T_anom_rcp26 = 1.5 * (years - 2020) / 80
T_anom_rcp45 = 2.5 * (years - 2020) / 80
T_anom_rcp85 = 4.5 * (years - 2020) / 80

# 基准气象条件 / Baseline meteorological conditions
T_base = 15.0  # °C
RH_base = 65.0
u2_base = 2.0
Rn_base = 12.0

# 计算不同情景下的ET / Calculate ET for different scenarios
scenarios = {
    'RCP2.6': {'co2': co2_rcp26, 'T_anom': T_anom_rcp26, 'color': 'green'},
    'RCP4.5': {'co2': co2_rcp45, 'T_anom': T_anom_rcp45, 'color': 'orange'},
    'RCP8.5': {'co2': co2_rcp85, 'T_anom': T_anom_rcp85, 'color': 'red'}
}

results = {}

for scenario_name, scenario_data in scenarios.items():
    co2 = scenario_data['co2']
    T_anom = scenario_data['T_anom']
    T = T_base + T_anom
    
    # 传统模型（只看温度效应）/ Traditional model (only temperature effect)
    et_trad = penman_monteith(
        temperature=T,
        relative_humidity=RH_base,
        wind_speed=u2_base,
        net_radiation=Rn_base
    )
    
    # CO2感知模型 / CO2-aware model
    et_co2 = pm_co2_aware(
        temperature=T,
        relative_humidity=RH_base,
        wind_speed=u2_base,
        net_radiation=Rn_base,
        co2=co2
    )
    
    results[scenario_name] = {
        'et_trad': et_trad,
        'et_co2': et_co2,
        'co2': co2,
        'T': T
    }

# 绘制四面板图 / Plot four-panel figure
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# 面板1：CO2浓度 / Panel 1: CO2 concentration
ax = axes[0, 0]
for scenario_name, scenario_data in scenarios.items():
    ax.plot(years, results[scenario_name]['co2'], linewidth=2.5, 
            color=scenario_data['color'], label=scenario_name, marker='o', markersize=5)
ax.set_ylabel('CO2 Concentration / CO2浓度 (ppm)', fontsize=11, fontweight='bold')
ax.set_title('A) Future CO2 Projections / 未来CO2预测', fontsize=13, fontweight='bold')
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3)

# 面板2：温度 / Panel 2: Temperature
ax = axes[0, 1]
for scenario_name, scenario_data in scenarios.items():
    ax.plot(years, results[scenario_name]['T'], linewidth=2.5, 
            color=scenario_data['color'], label=scenario_name, marker='s', markersize=5)
ax.set_ylabel('Temperature / 温度 (°C)', fontsize=11, fontweight='bold')
ax.set_title('B) Future Temperature Projections / 未来温度预测', fontsize=13, fontweight='bold')
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3)

# 面板3：传统模型ET / Panel 3: Traditional model ET
ax = axes[1, 0]
for scenario_name, scenario_data in scenarios.items():
    ax.plot(years, results[scenario_name]['et_trad'], linewidth=2.5, 
            color=scenario_data['color'], label=scenario_name, marker='^', markersize=5)
ax.set_ylabel('PET (mm/day)', fontsize=11, fontweight='bold')
ax.set_xlabel('Year / 年份', fontsize=11, fontweight='bold')
ax.set_title('C) Traditional PM: ET Increases with Warming\nC) 传统PM：ET随变暖增加', 
             fontsize=13, fontweight='bold')
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3)

# 面板4：CO2感知模型ET / Panel 4: CO2-aware model ET
ax = axes[1, 1]
for scenario_name, scenario_data in scenarios.items():
    ax.plot(years, results[scenario_name]['et_co2'], linewidth=2.5, 
            color=scenario_data['color'], label=scenario_name, marker='D', markersize=5)
ax.set_ylabel('PET (mm/day)', fontsize=11, fontweight='bold')
ax.set_xlabel('Year / 年份', fontsize=11, fontweight='bold')
ax.set_title('D) PM-CO2: Stomatal Effect Dampens ET Rise\nD) PM-CO2：气孔效应抑制ET增长', 
             fontsize=13, fontweight='bold')
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# 打印2100年预测 / Print 2100 projections
print("\n🔮 2100年预测 / Projections for 2100")
print("="*80)
print(f"{'Scenario':<10} {'CO2 (ppm)':<12} {'Temp (°C)':<12} {'ET_trad':<12} {'ET_CO2':<12} {'Diff (%)':<10}")
print("-"*80)

for scenario_name in ['RCP2.6', 'RCP4.5', 'RCP8.5']:
    co2_2100 = results[scenario_name]['co2'][-1]
    T_2100 = results[scenario_name]['T'][-1]
    et_trad_2100 = results[scenario_name]['et_trad'][-1]
    et_co2_2100 = results[scenario_name]['et_co2'][-1]
    diff_pct = (et_trad_2100 - et_co2_2100) / et_trad_2100 * 100
    
    print(f"{scenario_name:<10} {co2_2100:<12.0f} {T_2100:<12.1f} {et_trad_2100:<12.2f} "
          f"{et_co2_2100:<12.2f} {diff_pct:<10.1f}")

print("\n💡 重要结论 / Important Conclusions:")
print("  1. 传统模型预测ET持续上升（温室效应主导）")
print("     Traditional model predicts continuous ET rise (greenhouse effect dominates)")
print("\n  2. CO2感知模型显示ET增长被显著抑制（气孔效应部分抵消）")
print("     CO2-aware model shows significantly dampened ET rise (stomatal effect partially offsets)")
print("\n  3. 在RCP8.5情景下，两者差异高达20%以上！")
print("     Under RCP8.5, the difference exceeds 20%!")
````

## 🔥 实验3：温度vs.CO2的拉锯战 / Experiment 3: Temperature vs. CO2 Tug-of-War

让我们分离两种效应，看看哪个更强？ / Let's separate the two effects and see which is stronger?

````python
# 基准条件 / Baseline conditions
T_base = 15.0
CO2_base = 380.0
RH = 60.0
u2 = 2.5
Rn = 12.0

# 创建网格：温度 × CO2 / Create grid: Temperature × CO2
T_range = np.linspace(10, 25, 50)  # 10-25°C
CO2_range = np.linspace(280, 900, 50)  # 280-900 ppm

T_grid, CO2_grid = np.meshgrid(T_range, CO2_range)

# 计算ET网格 / Calculate ET grid
ET_grid = np.zeros_like(T_grid)

for i in range(len(CO2_range)):
    for j in range(len(T_range)):
        ET_grid[i, j] = pm_co2_aware(
            temperature=T_grid[i, j],
            relative_humidity=RH,
            wind_speed=u2,
            net_radiation=Rn,
            co2=CO2_grid[i, j]
        )

# 绘制等高线图 / Plot contour map
fig, ax = plt.subplots(figsize=(14, 10))

# 填充等高线 / Filled contours
contourf = ax.contourf(T_grid, CO2_grid, ET_grid, levels=20, cmap='RdYlBu_r', alpha=0.8)
contour = ax.contour(T_grid, CO2_grid, ET_grid, levels=10, colors='black', linewidths=0.5, alpha=0.3)
ax.clabel(contour, inline=True, fontsize=9, fmt='%.1f')

# 颜色条 / Colorbar
cbar = plt.colorbar(contourf, ax=ax)
cbar.set_label('PET (mm/day)', fontsize=12, fontweight='bold')

# 标注关键路径 / Annotate key pathways
# 路径1：只有温度变化 / Pathway 1: Temperature change only
T_path1 = np.linspace(15, 20, 10)
CO2_path1 = np.ones_like(T_path1) * 380
ax.plot(T_path1, CO2_path1, 'g-', linewidth=3, label='Pure warming\n(CO2=380 ppm)', marker='o')

# 路径2：只有CO2变化 / Pathway 2: CO2 change only
T_path2 = np.ones(10) * 15
CO2_path2 = np.linspace(380, 700, 10)
ax.plot(T_path2, CO2_path2, 'b-', linewidth=3, label='Pure CO2 rise\n(T=15°C)', marker='s')

# 路径3：RCP8.5轨迹 / Pathway 3: RCP8.5 trajectory
T_path3 = np.linspace(15, 19.5, 10)
CO2_path3 = np.linspace(380, 900, 10)
ax.plot(T_path3, CO2_path3, 'r-', linewidth=3, label='RCP8.5 trajectory\n(Both effects)', marker='^')

ax.set_xlabel('Temperature / 温度 (°C)', fontsize=13, fontweight='bold')
ax.set_ylabel('CO2 Concentration / CO2浓度 (ppm)', fontsize=13, fontweight='bold')
ax.set_title('The Battle: Temperature vs. CO2 Effects on ET\n温度与CO2对ET影响的较量', 
             fontsize=15, fontweight='bold')
ax.legend(fontsize=11, loc='upper left')
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# 计算三条路径的终点ET / Calculate endpoint ET for three pathways
et_warming_only = pm_co2_aware(20, RH, u2, Rn, 380)
et_co2_only = pm_co2_aware(15, RH, u2, Rn, 700)
et_combined = pm_co2_aware(19.5, RH, u2, Rn, 900)
et_baseline = pm_co2_aware(15, RH, u2, Rn, 380)

print("\n🎯 三种路径比较 / Three Pathway Comparison")
print("="*70)
print(f"基准 / Baseline (T=15°C, CO2=380 ppm):          ET = {et_baseline:.2f} mm/day")
print(f"\n路径1 / Pathway 1 (纯升温 / Pure warming):")
print(f"  终点 / Endpoint (T=20°C, CO2=380 ppm):        ET = {et_warming_only:.2f} mm/day")
print(f"  变化 / Change: +{(et_warming_only - et_baseline):.2f} mm/day ({(et_warming_only/et_baseline - 1)*100:.1f}%)")
print(f"\n路径2 / Pathway 2 (纯CO2升高 / Pure CO2 rise):")
print(f"  终点 / Endpoint (T=15°C, CO2=700 ppm):        ET = {et_co2_only:.2f} mm/day")
print(f"  变化 / Change: {(et_co2_only - et_baseline):.2f} mm/day ({(et_co2_only/et_baseline - 1)*100:.1f}%)")
print(f"\n路径3 / Pathway 3 (RCP8.5，两者兼有 / Both effects):")
print(f"  终点 / Endpoint (T=19.5°C, CO2=900 ppm):      ET = {et_combined:.2f} mm/day")
print(f"  变化 / Change: {(et_combined - et_baseline):.2f} mm/day ({(et_combined/et_baseline - 1)*100:.1f}%)")

print("\n💡 结论 / Conclusions:")
print("  ✅ 温度效应：促进ET增加")
print("     Temperature effect: Promotes ET increase")
print("  ❌ CO2效应：抑制ET增加")
print("     CO2 effect: Suppresses ET increase")
print("  ⚖️ 净效应取决于两者的相对强度！")
print("     Net effect depends on relative strength of both!")
````

## 📚 总结与思考 / Summary and Reflection

### 🎓 你学到了什么？ / What Did You Learn?

1. **蒸发悖论的根源 / Root of the Evaporation Paradox:**
   - 不仅是物理过程（温度、辐射）
   - 更重要的是生物响应（植物气孔调节）
   - Not just physical processes (temperature, radiation)
   - More importantly, biological responses (plant stomatal regulation)

2. **CO2的双重角色 / Dual Role of CO2:**
   - **温室气体**：加剧全球变暖 → 促进蒸发
   - **光合作用底物**：抑制气孔开度 → 抑制蒸发
   - **Greenhouse gas**: Intensifies warming → Promotes evaporation
   - **Photosynthesis substrate**: Suppresses stomatal opening → Suppresses evaporation

3. **模型的重要性 / Importance of Models:**
   - 传统模型可能严重高估未来ET
   - 必须在气候模式中考虑植物生理响应
   - Traditional models may seriously overestimate future ET
   - Must consider plant physiological responses in climate models

4. **不确定性 / Uncertainties:**
   - 不同植物物种的CO2响应差异很大
   - 养分限制、水分胁迫会改变响应
   - 需要更多实验数据验证
   - Different plant species respond differently to CO2
   - Nutrient limitation and water stress alter responses
   - More experimental data needed for validation

### 🤔 思考题 / Discussion Questions
