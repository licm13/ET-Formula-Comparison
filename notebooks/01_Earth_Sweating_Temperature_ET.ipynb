## 🤔 引言：地球的散热系统 / Introduction: Earth's Cooling System

### 人体类比 / Human Body Analogy

想象一下：
- 当你跑步后感到热的时候，身体会**出汗**来降温
- 汗水蒸发时会**吸收热量**，让皮肤凉下来
- 温度越高，出汗越多；风越大，蒸发越快

地球表面也是一样的！

- 🌲 植物会通过叶片**蒸腾**水分

Imagine:
- When you feel hot after running, your body **sweats** to cool down
- When sweat evaporates, it **absorbs heat**, cooling your skin
- Higher temperature → more sweat; stronger wind → faster evaporation

The Earth's surface works the same way!
- 🌊 Oceans, lakes, and soil **evaporate** water
- 🌲 Plants **transpire** water through their leaves
- Together, these processes are called **Evapotranspiration (ET)**

## 🔬 物理直觉：为什么温度重要？ / Physical Intuition: Why Does Temperature Matter?

### Clausius-Clapeyron 方程（简化版）

空气能"装"多少水汽？这取决于**饱和水汽压**（saturation vapor pressure, $e_s$）。

$$
e_s(T) = 0.6108 \times e^{\frac{17.27 T}{T + 237.3}}
$$

其中：
- $T$ 是温度（°C）
- $e_s$ 是饱和水汽压（kPa）

**关键发现：**
- 温度每升高 10°C，饱和水汽压**几乎翻倍**！
- 这就是为什么夏天比冬天蒸发快得多

## 💻 代码实战：从零开始计算饱和水汽压
## 💻 Hands-On Coding: Calculate Saturation Vapor Pressure from Scratch

```python
import numpy as np
import matplotlib.pyplot as plt
import sys
import os

# 设置中文字体（如果可用）/ Set Chinese-capable font if available
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei', 'Arial', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

# 添加项目路径 / Add project path
sys.path.insert(0, os.path.abspath('..'))

print("✅ Libraries imported successfully!")
print("✅ 库导入成功！")
```

### 步骤1：手动实现饱和水汽压公式 / Step 1: Manually Implement the Formula

```python
def saturation_vapor_pressure_manual(T):
    """
    计算饱和水汽压（Tetens 公式）
    Calculate saturation vapor pressure (Tetens formula)
    
    参数 / Parameters:
        T: 温度（°C） / Temperature (°C)
    
    返回 / Returns:
        e_s: 饱和水汽压（kPa） / Saturation vapor pressure (kPa)
    """
    return 0.6108 * np.exp((17.27 * T) / (T + 237.3))

# 测试一下 / Test it
T_test = 20.0  # 20°C
es_test = saturation_vapor_pressure_manual(T_test)

print(f"🌡️ 温度 / Temperature: {T_test}°C")
print(f"💧 饱和水汽压 / Saturation vapor pressure: {es_test:.4f} kPa")
print(f"\n💡 这意味着在 {T_test}°C 时，空气最多能容纳相当于 {es_test:.4f} kPa 压力的水汽！")
print(f"💡 This means at {T_test}°C, air can hold water vapor equivalent to {es_test:.4f} kPa pressure!")
```

### 步骤2：可视化温度-饱和水汽压曲线 / Step 2: Visualize the Temperature-Vapor Pressure Curve

```python
# 创建温度范围 / Create temperature range
temperatures = np.linspace(-10, 40, 100)  # -10°C to 40°C
vapor_pressures = saturation_vapor_pressure_manual(temperatures)

# 绘图 / Plot
plt.figure(figsize=(10, 6))
plt.plot(temperatures, vapor_pressures, linewidth=2.5, color='steelblue')
plt.xlabel('Temperature / 温度 (°C)', fontsize=12)
plt.ylabel('Saturation Vapor Pressure / 饱和水汽压 (kPa)', fontsize=12)
plt.title('Clausius-Clapeyron Relationship: How Air "Holds" Water Vapor\n温度与饱和水汽压的关系', 
          fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)

# 标注关键点 / Annotate key points
for T_mark in [0, 10, 20, 30]:
    es_mark = saturation_vapor_pressure_manual(T_mark)
    plt.scatter(T_mark, es_mark, s=100, color='red', zorder=5)
    plt.text(T_mark, es_mark + 0.3, f'{es_mark:.2f} kPa', 
             ha='center', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()

print("\n🔍 观察规律 / Key Observations:")
print(f"  • 0°C → 10°C: 饱和水汽压增加 {saturation_vapor_pressure_manual(10)/saturation_vapor_pressure_manual(0):.2f} 倍")
print(f"  • 10°C → 20°C: 饱和水汽压增加 {saturation_vapor_pressure_manual(20)/saturation_vapor_pressure_manual(10):.2f} 倍")
print(f"  • 20°C → 30°C: 饱和水汽压增加 {saturation_vapor_pressure_manual(30)/saturation_vapor_pressure_manual(20):.2f} 倍")
print("\n💡 结论：温度越高，空气'装'水的能力呈**指数增长**！")
print("💡 Conclusion: The capacity of air to 'hold' water vapor grows **exponentially** with temperature!")
```

## 🌾 第一个真正的ET公式：Hargreaves 方法
## 🌾 Your First Real ET Formula: The Hargreaves Method

### 背景故事 / Background Story

1985年，两位科学家 George Hargreaves 和 Zohrab Samani 提出了一个简单但强大的公式：

**只需要温度数据，就能估算蒸散发！**

In 1985, two scientists George Hargreaves and Zohrab Samani proposed a simple yet powerful formula:

**Estimate evapotranspiration using only temperature data!**

### 公式 / Formula

$$
\text{PET} = 0.0023 \times R_a \times (T_{\text{mean}} + 17.8) \times \sqrt{T_{\text{max}} - T_{\text{min}}}
$$

其中 / Where:
- $R_a$: 天文辐射（地球在该位置能接收到的太阳辐射）/ Extraterrestrial radiation
- $T_{\text{mean}}$: 日平均温度 / Mean daily temperature (°C)
- $T_{\text{max}}$: 日最高温度 / Maximum daily temperature (°C)
- $T_{\text{min}}$: 日最低温度 / Minimum daily temperature (°C)

### 物理意义 / Physical Meaning

1. **能量供应** ($R_a$): 太阳提供蒸发所需的能量
2. **温度水平** ($T_{\text{mean}} + 17.8$): 控制饱和水汽压
3. **昼夜温差** ($\sqrt{T_{\text{max}} - T_{\text{min}}}$): 反映大气的干燥程度

1. **Energy supply** ($R_a$): Sun provides energy for evaporation
2. **Temperature level** ($T_{\text{mean}} + 17.8$): Controls saturation vapor pressure
3. **Diurnal temperature range** ($\sqrt{T_{\text{max}} - T_{\text{min}}}$): Reflects atmospheric dryness

**神奇之处：** 温差大 → 空气干燥 → 蒸发快！

**The magic:** Large temperature range → Dry air → Fast evaporation!

### 动手计算：北京夏天的蒸散发 / Calculate: Summer ET in Beijing

```python
from pet_comparison.formulas import hargreaves

# 北京夏季典型天气 / Typical summer weather in Beijing
T_mean = 28.0   # °C
T_max = 35.0    # °C
T_min = 21.0    # °C
doy = 180       # 第180天（6月底）/ Day 180 (late June)
latitude = 39.9 # 北京纬度 / Beijing latitude

# 计算PET / Calculate PET
pet_beijing = hargreaves(
    temperature_mean=T_mean,
    temperature_max=T_max,
    temperature_min=T_min,
    doy=doy,
    latitude=latitude
)

print("🌏 北京夏季蒸散发 / Summer ET in Beijing")
print("="*50)
print(f"日期 / Date: 第 {doy} 天（6月29日）/ Day {doy} (June 29)")
print(f"平均温度 / Mean temperature: {T_mean}°C")
print(f"最高温度 / Max temperature: {T_max}°C")
print(f"最低温度 / Min temperature: {T_min}°C")
print(f"昼夜温差 / Temperature range: {T_max - T_min}°C")
print(f"\n💧 蒸散发 / PET: {pet_beijing:.2f} mm/day")
print(f"\n💡 这意味着在这种天气下，地表每天会失去约 {pet_beijing:.2f} 毫米的水！")
print(f"💡 This means the surface loses about {pet_beijing:.2f} mm of water per day under these conditions!")
print(f"\n📊 如果这样持续30天，总蒸散发量 = {pet_beijing * 30:.1f} mm")
print(f"📊 If this continues for 30 days, total ET = {pet_beijing * 30:.1f} mm")
```

### 🧪 实验：控制变量法探索
### 🧪 Experiment: Control Variable Method

让我们改变一个变量，观察ET如何变化 / Let's change one variable and observe how ET changes

```python
# 实验1：温度的影响 / Experiment 1: Effect of temperature
print("🔬 实验1：温度如何影响ET？ / Experiment 1: How does temperature affect ET?")
print("="*70)

temperatures = np.array([10, 15, 20, 25, 30, 35])  # °C
pet_values = []

for T in temperatures:
    pet = hargreaves(
        temperature_mean=T,
        temperature_max=T + 7,
        temperature_min=T - 7,
        doy=180,
        latitude=40
    )
    pet_values.append(pet)
    print(f"  温度 / Temperature: {T:2.0f}°C  →  PET: {pet:5.2f} mm/day")

pet_values = np.array(pet_values)

# 可视化 / Visualize
plt.figure(figsize=(10, 6))
plt.plot(temperatures, pet_values, marker='o', linewidth=2.5, 
         markersize=10, color='orangered', label='Hargreaves PET')
plt.xlabel('Mean Temperature / 平均温度 (°C)', fontsize=12)
plt.ylabel('PET (mm/day)', fontsize=12)
plt.title('Temperature vs. Evapotranspiration\n温度对蒸散发的影响', 
          fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()

print(f"\n📈 从10°C到35°C，PET增加了 {pet_values[-1]/pet_values[0]:.2f} 倍！")
print(f"📈 From 10°C to 35°C, PET increased by {pet_values[-1]/pet_values[0]:.2f} times!")
```

```python
# 实验2：昼夜温差的影响 / Experiment 2: Effect of diurnal temperature range
print("\n🔬 实验2：昼夜温差如何影响ET？ / Experiment 2: How does temperature range affect ET?")
print("="*70)

temp_ranges = np.array([2, 5, 8, 10, 12, 15, 20])  # °C
pet_values_range = []

T_mean_fixed = 25.0  # 固定平均温度 / Fixed mean temperature

for dt in temp_ranges:
    pet = hargreaves(
        temperature_mean=T_mean_fixed,
        temperature_max=T_mean_fixed + dt/2,
        temperature_min=T_mean_fixed - dt/2,
        doy=180,
        latitude=40
    )
    pet_values_range.append(pet)
    print(f"  温差 / Range: {dt:2.0f}°C  →  PET: {pet:5.2f} mm/day")

pet_values_range = np.array(pet_values_range)

# 可视化 / Visualize
plt.figure(figsize=(10, 6))
plt.plot(temp_ranges, pet_values_range, marker='s', linewidth=2.5, 
         markersize=10, color='forestgreen', label='Hargreaves PET')
plt.xlabel('Diurnal Temperature Range / 昼夜温差 (°C)', fontsize=12)
plt.ylabel('PET (mm/day)', fontsize=12)
plt.title('Temperature Range vs. Evapotranspiration\n昼夜温差对蒸散发的影响', 
          fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()

print(f"\n🌡️ 物理解释：温差大 → 空气更干燥 → 蒸发更快")
print(f"🌡️ Physical explanation: Large range → Drier air → Faster evaporation")
print(f"\n💡 在沙漠地区（温差大），即使平均温度相同，ET也会更高！")
print(f"💡 In desert regions (large temperature range), ET is higher even with the same mean temperature!")
```

## 🌏 真实应用：模拟一年的蒸散发
## 🌏 Real Application: Simulate One Year of ET

让我们模拟一个地点一整年的ET变化 / Let's simulate one full year of ET for a location

```python
import pandas as pd

# 生成一年的合成气象数据 / Generate synthetic meteorological data for one year
days = np.arange(1, 366)  # Day of year 1-365
latitude_test = 40.0  # 北纬40度（约北京、伊斯坦布尔、纽约）/ 40°N (Beijing, Istanbul, New York)

# 模拟温度的季节变化 / Simulate seasonal temperature variation
T_mean_annual = 15.0  # 年平均温度 / Annual mean temperature
T_amplitude = 15.0    # 振幅 / Amplitude

T_mean_year = T_mean_annual + T_amplitude * np.sin(2 * np.pi * (days - 80) / 365)
T_max_year = T_mean_year + 7  # 最高温度 / Max temperature
T_min_year = T_mean_year - 7  # 最低温度 / Min temperature

# 计算全年的PET / Calculate PET for the entire year
pet_year = hargreaves(
    temperature_mean=T_mean_year,
    temperature_max=T_max_year,
    temperature_min=T_min_year,
    doy=days,
    latitude=latitude_test
)

# 创建DataFrame / Create DataFrame
df_year = pd.DataFrame({
    'Day': days,
    'T_mean': T_mean_year,
    'T_max': T_max_year,
    'T_min': T_min_year,
    'PET': pet_year
})

# 添加月份标签 / Add month labels
month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 
               'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
month_days = [15, 45, 75, 105, 135, 165, 195, 225, 255, 285, 315, 345]

# 绘制完整的年度图 / Plot full annual cycle
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 10))

# 上图：温度 / Top: Temperature
ax1.plot(days, T_mean_year, linewidth=2, color='red', label='Mean Temperature')
ax1.fill_between(days, T_min_year, T_max_year, alpha=0.3, color='orange', 
                  label='Temperature Range')
ax1.set_ylabel('Temperature / 温度 (°C)', fontsize=12, fontweight='bold')
ax1.set_title('Annual Temperature Cycle / 年度温度变化', fontsize=14, fontweight='bold')
ax1.legend(fontsize=11)
ax1.grid(True, alpha=0.3)
ax1.set_xticks(month_days)
ax1.set_xticklabels(month_names)

# 下图：PET / Bottom: PET
ax2.plot(days, pet_year, linewidth=2.5, color='blue', label='PET (Hargreaves)')
ax2.fill_between(days, 0, pet_year, alpha=0.3, color='skyblue')
ax2.set_ylabel('PET (mm/day)', fontsize=12, fontweight='bold')
ax2.set_xlabel('Month / 月份', fontsize=12, fontweight='bold')
ax2.set_title('Annual Evapotranspiration Cycle / 年度蒸散发变化', fontsize=14, fontweight='bold')
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3)
ax2.set_xticks(month_days)
ax2.set_xticklabels(month_names)

plt.tight_layout()
plt.show()

# 统计摘要 / Statistical summary
print("\n📊 年度统计摘要 / Annual Statistical Summary")
print("="*60)
print(f"平均温度 / Mean Temperature: {T_mean_year.mean():.1f}°C")
print(f"最高温度 / Max Temperature: {T_max_year.max():.1f}°C (第 {days[T_max_year.argmax()]} 天)")
print(f"最低温度 / Min Temperature: {T_min_year.min():.1f}°C (第 {days[T_min_year.argmin()]} 天)")
print(f"\n平均日PET / Mean Daily PET: {pet_year.mean():.2f} mm/day")
print(f"最大日PET / Max Daily PET: {pet_year.max():.2f} mm/day (第 {days[pet_year.argmax()]} 天)")
print(f"最小日PET / Min Daily PET: {pet_year.min():.2f} mm/day (第 {days[pet_year.argmin()]} 天)")
print(f"\n年总蒸散发 / Annual Total ET: {pet_year.sum():.0f} mm")
print(f"这相当于 {pet_year.sum()/1000:.2f} 米的水层！")
print(f"This is equivalent to a {pet_year.sum()/1000:.2f} meter layer of water!")
```
